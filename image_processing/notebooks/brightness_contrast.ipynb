# Task 5: Thực hành xử lý ảnh - Brightness và Contrast

Notebook này minh họa quá trình điều chỉnh độ sáng (brightness) và độ tương phản (contrast).

```python
import cv2
import matplotlib.pyplot as plt
import sys
import os

sys.path.append(os.path.abspath('image_processing/opencv_basics'))
from brightness_contrast import adjust_brightness, adjust_contrast
from io_basics import load_image

img_path = 'image_processing/input/sample.jpg'
img_bgr = load_image(img_path)

# Apply transformations
img_dark = adjust_brightness(img_bgr, -50)
img_bright = adjust_brightness(img_bgr, 50)
img_low_contrast = adjust_contrast(img_bgr, 0.5)
img_high_contrast = adjust_contrast(img_bgr, 1.5)
img_combined = adjust_brightness(adjust_contrast(img_bgr, 1.2), 30)

def to_rgb(img):
    return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
fig, axs = plt.subplots(2, 3, figsize=(15, 10))

axs[0, 0].imshow(to_rgb(img_bgr))
axs[0, 0].set_title('Original')
axs[0, 0].axis('off')

axs[0, 1].imshow(to_rgb(img_dark))
axs[0, 1].set_title('Dark (Brightness -50)')
axs[0, 1].axis('off')

axs[0, 2].imshow(to_rgb(img_bright))
axs[0, 2].set_title('Bright (Brightness +50)')
axs[0, 2].axis('off')

axs[1, 0].imshow(to_rgb(img_low_contrast))
axs[1, 0].set_title('Low Contrast (alpha=0.5)')
axs[1, 0].axis('off')

axs[1, 1].imshow(to_rgb(img_high_contrast))
axs[1, 1].set_title('High Contrast (alpha=1.5)')
axs[1, 1].axis('off')

axs[1, 2].imshow(to_rgb(img_combined))
axs[1, 2].set_title('Combined (alpha=1.2, beta=30)')
axs[1, 2].axis('off')

plt.tight_layout()
plt.show()
```
