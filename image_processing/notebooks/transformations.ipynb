# Task 4: Thực hành xử lý ảnh - Resize, Rotate và Flip

Notebook này minh họa quá trình sử dụng OpenCV để thay đổi kích thước, xoay và lật ảnh.

```python
import cv2
import matplotlib.pyplot as plt
import sys
import os

# Thêm đường dẫn để import custom modules
sys.path.append(os.path.abspath('image_processing/opencv_basics'))
from transformations import resize_image, rotate_image, flip_image
from io_basics import load_image

img_path = 'image_processing/input/sample.jpg'
img_bgr = load_image(img_path)
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# Apply transformations
img_resized = resize_image(img_bgr, 256, 256)
img_rot90 = rotate_image(img_bgr, 90)
img_rot180 = rotate_image(img_bgr, 180)
img_flip_h = flip_image(img_bgr, 'horizontal')
img_flip_v = flip_image(img_bgr, 'vertical')

def to_rgb(img):
    return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
fig, axs = plt.subplots(2, 3, figsize=(15, 10))

axs[0, 0].imshow(img_rgb)
axs[0, 0].set_title(f'Original {img_rgb.shape}')
axs[0, 0].axis('off')

axs[0, 1].imshow(to_rgb(img_resized))
axs[0, 1].set_title(f'Resize 256x256 {img_resized.shape}')
axs[0, 1].axis('off')

axs[0, 2].imshow(to_rgb(img_rot90))
axs[0, 2].set_title(f'Rotate 90 {img_rot90.shape}')
axs[0, 2].axis('off')

axs[1, 0].imshow(to_rgb(img_rot180))
axs[1, 0].set_title(f'Rotate 180 {img_rot180.shape}')
axs[1, 0].axis('off')

axs[1, 1].imshow(to_rgb(img_flip_h))
axs[1, 1].set_title(f'Flip Horizontal {img_flip_h.shape}')
axs[1, 1].axis('off')

axs[1, 2].imshow(to_rgb(img_flip_v))
axs[1, 2].set_title(f'Flip Vertical {img_flip_v.shape}')
axs[1, 2].axis('off')

plt.tight_layout()
plt.show()
```
