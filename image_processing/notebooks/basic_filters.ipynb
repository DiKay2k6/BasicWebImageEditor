# Task 3: Thực hành xử lý ảnh - Grayscale và Blur

Notebook này minh họa quá trình sử dụng thư viện OpenCV để đọc ảnh, áp dụng bộ lọc (Grayscale, Blur) và hiển thị kết quả bằng Matplotlib.

```python
import cv2
import matplotlib.pyplot as plt
import sys
import os

# Thêm đường dẫn tới thư mục opencv_basics để import các hàm dùng chung
sys.path.append(os.path.abspath('image_processing/opencv_basics'))
from basic_filters import apply_grayscale, apply_blur
from io_basics import load_image

# 1. Đọc ảnh đầu vào
img_path = 'image_processing/input/sample.jpg'
img_bgr = load_image(img_path)

# OpenCV dùng định dạng BGR, trong khi Matplotlib dùng RGB
# Cần chuyển đổi để hiển thị màu sắc chính xác trên Notebook
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
```

```python
# 2. Áp dụng hiệu ứng
img_gray = apply_grayscale(img_bgr)
img_blur_bgr = apply_blur(img_bgr, kernel_size=(15, 15))

# Chuyển đổi màu cho ảnh blur để hiển thị
img_blur_rgb = cv2.cvtColor(img_blur_bgr, cv2.COLOR_BGR2RGB)
```

```python
# 3. Hiển thị kết quả so sánh
plt.figure(figsize=(18, 6))

# Ảnh gốc
plt.subplot(1, 3, 1)
plt.title('Ảnh Gốc (Original)')
plt.imshow(img_rgb)
plt.axis('off')

# Ảnh Grayscale
plt.subplot(1, 3, 2)
plt.title('Ảnh Grayscale')
plt.imshow(img_gray, cmap='gray')
plt.axis('off')

# Ảnh Blur
plt.subplot(1, 3, 3)
plt.title('Ảnh Gaussian Blur')
plt.imshow(img_blur_rgb)
plt.axis('off')

plt.show()
```
